```python
import os
import shutil
import random

random.seed(42)

# ORIGINAL dataset (oyā images thiyena folder)
SOURCE_DIR = r"D:\predictions\Handwriting\dataset"

# TARGET split folder (new)
TARGET_DIR = r"D:\predictions\Handwriting\dataset_split"

CLASSES = ["Healthy", "Parkinson"]

TRAIN_RATIO = 0.8
VAL_RATIO = 0.1
TEST_RATIO = 0.1
```

```python
for cls in CLASSES:
    src_class_dir = os.path.join(SOURCE_DIR, cls)
    images = os.listdir(src_class_dir)
    random.shuffle(images)

    total = len(images)
    train_end = int(total * TRAIN_RATIO)
    val_end = train_end + int(total * VAL_RATIO)

    splits = {
        "train": images[:train_end],
        "val": images[train_end:val_end],
        "test": images[val_end:]
    }

    for split, split_images in splits.items():
        dst_dir = os.path.join(TARGET_DIR, split, cls)
        os.makedirs(dst_dir, exist_ok=True)

        for img in split_images:
            shutil.copy(
                os.path.join(src_class_dir, img),
                os.path.join(dst_dir, img)
            )

    print(f"{cls}: Train={len(splits['train'])}, Val={len(splits['val'])}, Test={len(splits['test'])}")
```
